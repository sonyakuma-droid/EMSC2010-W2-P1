## Random numbers
Below is some simple code to *generate uniformly* distributed random numbers. Don't worry about how the code works, we're just going to use it to demonstrate a concept.



````python
class SimpleRNG:
    """
    A Linear Congruential Generator (LCG).
    Formula: next = (a * current + c) % m
    """
    def __init__(self, seed=42):
        self.state = seed
        self.a = 1664525      # Multiplier
        self.c = 1013904223   # Increment
        self.m = 2**32        # Modulus

    def next_int(self):
        """Generate the next raw integer."""
        self.state = (self.a * self.state + self.c) % self.m
        return self.state

    def next_float(self):
        """Generate a float in [0, 1)."""
        return self.next_int() / self.m
````

Generate 5 random numbers using a *seed* and print them to the screen

````python
rng1 = SimpleRNG(seed=99)
for _ in range(5):
    print(f"  rng1: {rng1.next_float():.6f}")
````

### Using Numpy
Rather than writing our own code, we typically use a package like ```numpy``` to generate random numbers.

````python
import numpy as np

# Create a generator with a fixed seed
rng = np.random.default_rng(seed=99)

# Generate an array of 5 numbers in [0, 1)
print(rng.uniform(size=5))

# Generate numbers in a custom range, e.g. [10, 20)
print(rng.uniform(low=10, high=20, size=5))

### statistics ###
import numpy as np
import scipy.stats as stats
import statsmodels.stats.api as sms

# Sample data
data = [12, 15, 14, 10, 8, 16, 14, 15, 13, 11]

# 1. --- Calculate 95% Confidence Interval of the Mean (Method A: scipy) ---
mean = np.mean(data)
se = stats.sem(data)  # Standard error
ci_scipy = stats.t.interval(0.95, len(data)-1, loc=mean, scale=se)
print(f"95% CI (Scipy): {ci_scipy}")

# 2. --- Calculate 95% Confidence Interval of the Mean (Method B: statsmodels) ---
cm = sms.DescrStatsW(data)
ci_sm = cm.tconfint_mean(alpha=0.05)
print(f"95% CI (Statsmodels): {ci_sm}")

# 3. --- Calculate Probability (e.g., P(X < 14) for normal distribution) ---
# Assuming sample mean/std represent population
prob = stats.norm.cdf(14, loc=np.mean(data), scale=np.std(data, ddof=1))
print(f"Probability P(X < 14): {prob:.4f}")
````
